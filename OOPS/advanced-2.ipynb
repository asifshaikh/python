Association -> its a connection between two or more classes that shows how their objects are related to each other \
Simple Terms : Class A knows about Class B and can use it

````python
# Association Example
class Teacher:
    def __init__(self, name):
        self.name = name
        self.students = []

    def add_student(self, student):
        self.students.append(student)

    def get_students(self):
        return [student.name for student in self.students]

class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name

# Usage
teacher = Teacher("Dr. Smith")
student1 = Student("Alice")
student2 = Student("Bob")

teacher.add_student(student1)
teacher.add_student(student2)

print(f"Teacher: {teacher.name}")
print(f"Students: {', '.join(teacher.get_students())}")
````

Aggregation -> is a weak `has-a` relationship, where one class contains the object of another class, but the contained object can survive independently even if the container is destroyed \

Simple Terms: Class A contains Class B, but B can survive without A

````python
class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    def get_name(self):
        return self.__name

    def get_roll_number(self):
        return self.__roll_number

class Department:
    def __init__(self,dept_name):
        self.__dept_name = dept_name
        self.__students = []

    def add_student(self, student):
        self.__students.append(student)

    def get_students(self):
        for student in self.__students:
            print(f"Name: {student.get_name()}, Roll Number: {student.get_roll_number()}")
# Usage
student1 = Student("Alice", 101)
student2 = Student("Bob", 102)

department = Department("Computer Science")

department.add_student(student1)
department.add_student(student2)


department.get_students()
del department
print(f"Students still exists: {student1.get_name()}, {student2.get_name()}")
````

Composition -> is a strong `has-a` relationship where one class owns the object of another class. If the container is destroyed the object is also destroyed \

Simple Terms: Class A owns Class B completely, If A dies B also dies

````python
# Composition Example
class Engine:
    def __init__(self, horsepower, type):
        self.__horsepower = horsepower
        self.__type = type

    def get_details(self):
        print(f"{self.__horsepower} HP, {self.__type}")

    def start(self):
        print(f"{self.__type} engine started with {self.__horsepower} HP.") 

class Car:
    def __init__(self, make, model, horsepower, type):
        self.__make = make
        self.__model = model
        self.__engine = Engine(horsepower, type)

    def get_details(self):
        print(f"Car: {self.__make} {self.__model}")
        self.__engine.get_details()

    def start_engine(self):
        self.__engine.start()

# Usage
car = Car("Toyota", "Camry", 200, "V6")

car.get_details()
car.start_engine()

````

Inheritance -> defines an `is-a` relationship where a child inherits all the methods and properties of parent \

Simple Terms: Child class is a type of Parent class

````python
# Inheritance Example
class Animal:
    def __init__(self,name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating.")

class Dog(Animal):
    def __init__(self,name, breed):
        super().__init__(name)
        self.breed = breed

    def bark(self):
        print(f"{self.name} is barking.")
# Usage
dog = Dog("Buddy", "Labrador")
dog.eat()
dog.bark()
````

Dataclass:
- Python 3.7 introduced dataclass \
    ``` from dataclasses import dataclass ```
- It automatically generates:
    - `__init__`
    - `__repr__`
    - `__eq__`
    - and many more

````python
from dataclasses import dataclass

@dataclass
class Person:
    name: str
    age: int

person = Person("John", 30)
print(person)
````

Shallow Copy:
- `import copy`
- `copy.copy()`
- copies only outer object
- nested objects are shared
- original object changes

````python
import copy

a = [[1, 2], [3, 4]]
print("Original:", a)
b = copy.copy(a)
b[0][0] = 99
print("Shallow copy:", a)
````

Deep copy:
- copies everything recursively

````python
import copy

a=[[1,2],[3,4]]

b=copy.deepcopy(a)

b[0][0]=99

print(a)
````
