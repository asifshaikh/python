Instance Method:
- work on instance variables
- always receive `self` as the first argument
- used when we need object specific data

Class Method:
- work on class variables
- always receive `cls` as the first argument

Static Method:
- they do not take `cls` or `self` as argument
- they work normally like utility function but belong to the class

```python
# Instance Method Example
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        return f"Name: {self.name}, Age: {self.age}"

student1 = Student("John", 20)
print(student1.display_info())
```

```python
# Class Method Example
class Student:
    school_name = "ABC School"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def get_school_name(cls):
        return cls.school_name

    @classmethod
    def set_school_name(cls, name):
        cls.school_name = name

student1 = Student("John", 20)
print(student1.get_school_name())

student1.set_school_name("XYZ School")
print(student1.get_school_name())
```

```python
# Static Method Example
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @staticmethod
    def is_adult(age):
        return age >= 18

student1 = Student("John", 20)
print(student1.is_adult(20))
```

```python
# Example combining all three
class Student:
    school_name = "ABC School"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        return f"Name: {self.name}, Age: {self.age}"

    @classmethod
    def get_school_name(cls):
        return cls.school_name

    @classmethod
    def set_school_name(cls, name):
        cls.school_name = name

    @staticmethod
    def is_adult(age):
        return age >= 18

student1 = Student("John", 20)
print(student1.display_info())
print(student1.get_school_name())
student1.set_school_name("XYZ School")
print(student1.get_school_name())
print(student1.is_adult(20))
```

Magic / Dunder Methods:
- They begin and end with double underscore
- Ex: `__init__` which mean initialize object

More common Magic Methods:
- `__str__` -> User friendly string representation
- `__repr__` -> Developer/debug representation
- `__len__` -> supports `len()`
- `__new__` -> creates the object in memory and runs before `__init__`

```python
# Example without __str__ method
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("John", 20)
print(student1)
```

```python
# Example with __str__ method
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Name: {self.name}, Age: {self.age}"

student1 = Student("John", 20)
print(student1)
```

```python
# Example of __repr__ method
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Name: {self.name}, Age: {self.age}"

    def __repr__(self):
        return f"Student('{self.name}', {self.age})"

student1 = Student("John", 20)
print(student1)
print(repr(student1))
```

```python
# Example of __new__ method
class Demo:

    def __new__(cls):
        print("Creating Object")
        return super().__new__(cls)

    def __init__(self):
        print("Initializing")

Demo()
```
