Inheritance -> It allows a subclass(child) to inherit properties and methods of another class(parent)

Types of Inheritance:
- Single Inheritance / One child and one parent / class B(A)
- Multilevel Inheritance / Child -> Parent -> Grandparent / class C(B)
- Multiple Inheritance / One child and multiple parents / class B(C,A)
- Hierarchical Inheritance / One parent and multiple children / class B(A), class C(A)
- Hybrid Inheritance / combination of above types / complex real world use cases

super() -> It is used to call parent class methods inside the child class

Python supports multiple Inheritance using Method Resolution Order (MRO) \
MRO follows left to right order in the parent list

```python
# Single Inheritance
class Parent:
    def func1(self):
        print("This function is in parent class.")
        
class Child(Parent):
    def func2(self):
        print("This function is in child class.")
        
c = Child()
c.func1()
c.func2()
```

```python
# Multilevel Inheritance
class Grandparent:
    def func1(self):
        print("This function is in grandparent class.")
        
class Parent(Grandparent):
    def func2(self):
        print("This function is in parent class.")
        
class Child(Parent):
    def func3(self):
        print("This function is in child class.")
        
c = Child()
c.func1()
c.func2()
```

```python
# Multiple Inheritance
class Mother:
    mothername = ""
    
    def mother(self):
        print(self.mothername)
        
class Father:
    fathername = ""
    
    def father(self):
        print(self.fathername)
        
class Son(Mother, Father):
    def parents(self):
        print("Father :", self.fathername)
        print("Mother :", self.mothername)
        
s1 = Son()
s1.fathername = "Asif"
s1.mothername = "Namrah"
s1.parents()
```

```python
# Hierarchical Inheritance
class Parent:
    def func1(self):
        print("This function is in parent class.")

class Child1(Parent):
    def func2(self):
        print("This function is in child 1.")

class Child2(Parent):
    def func3(self):
        print("This function is in child 2.")

c1 = Child1()
c2 = Child2()
c1.func1()
c1.func2()
c2.func1()
c2.func3()
```

```python
# Multiple Inheritance and 
class A:
    def process(self): print("Class A")

class B(A):
    def process(self): print("Class B")

class C(A):
    def process(self): print("Class C")

class D(B, C):
    pass

obj = D()
obj.process()
print(D.mro())  # Method Resolution Order
```

Objective: Implement a simple inheritance hierarchy.

Classes:

Person → base class (name, age)
Teacher → inherits Person (subject)
Student → inherits Person (marks)
School → manages lists of students & teachers

Tasks:

Display teacher & student details.
Add a method to print overall school summary.
Use super() in constructors.
Expected Output Example:

Teacher: Mr. Sharma, Subject: Math
Student: Asif, Marks: 92
Total Teachers: 1, Total Students: 1

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        
class Teacher(Person):
    teacher_count = 0
    def __init__(self,name,age,subject):
        super().__init__(name,age)
        self.subject = subject
        Teacher.teacher_count += 1

    def display(self):
        print("Teacher:", self.name, "Subject:", self.subject)

class Student(Person):
    student_count = 0
    def __init__(self,name,age,marks):
        super().__init__(name,age)
        self.marks = marks
        Student.student_count += 1

    def display(self):
        print("Student:", self.name, "Marks:", self.marks)

class School:
    def display_counts(self):
        print("Teacher count:", Teacher.teacher_count)
        print("Student count:", Student.student_count)


teacher = Teacher("John", 30, "Maths")
student = Student("Jane", 20, 85)
teacher1 = Teacher("Alice", 35, "Science")
student1 = Student("Bob", 25, 92)

teacher.display()
student.display()
teacher1.display()
student1.display()

school = School()
school.display_counts()
```
