OOPS -> It is a way of writing programs by creating objects that include attributes (data/properties) and methods (behaviour/actions) \
Class -> It is a blueprint of the object \
Object -> It is an instance of the class

```python
class Student: # creating a class   
    def __init__(self,name,age): # __init_ is a constructor that initializes the attributes of the class
        self.name = name # self is a keyword that refers to the instance of the class
        self.age = age

    def print_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")

student1 = Student("John", 20) # creating an object
student1.print_info()
```

Instance Variables:
- Belongs to the object (each object gets its own copy)
- created using self

Class Variables:
- shared by all objects of a class
- declared outside __init__() at class level

```python
class Student:
    school_name = "ABC School" # class variable

    def __init__(self, name, age):
        self.name = name # instance variable
        self.age = age

    def print_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"School Name: {Student.school_name}") # accessing class variable

student1 = Student("John", 20)
student1.print_info()

student2 = Student("Jane", 19)
student2.print_info()

Student.school_name = "XYZ School" # changing class variable
student1.print_info()
student2.print_info()
```
