#### Functional Requirement Question

Design and implement a Python library management system that includes:

- a `Book` class with fields:
  - `title`
  - `author`
  - `available`

- a `Member` class with fields:
  - `name`
  - `borrowed_books`

- a `Library` class that uses composition to manage multiple books and provides the following methods:
  - `add_book()`
  - `remove_book()`
  - `search_book()`
  - `borrow_book()`
  - `return_book()`
  - `display_books()`

```python
class Book:
    def __init__(self,title,author,available):
        self.title = title
        self.author = author
        self.available = available

class Member:
    def __init__(self,name):
        self.name = name
        self.borrowed_books = []

class Library:
    def __init__(self,title,author,available):
        self.books = [Book(title,author,available)]

    def add_book(self,title,author,available):
        self.books.append(Book(title,author,available))

    def remove_book(self,title):
        for book in self.books:
            if book.title == title:
                self.books.remove(book)
                return True
        return False

    def search_book(self,title):
        for book in self.books:
            if book.title == title:
                return book
        return None

    def borrow_book(self,title,member):
        book = self.search_book(title)
        if book and book.available:
            book.available = False
            member.borrowed_books.append(book)
            return True
        return False

    def return_book(self,title,member):
        book = self.search_book(title)
        if book and not book.available:
            book.available = True
            member.borrowed_books.remove(book)
            return True
        return False

    def display_books(self):
        print("Available Books:")
        for book in self.books:
            if book.available:
                print(f"Title: {book.title}, Author: {book.author}, Available: Yes")
            

# Usage
library = Library("The Great Gatsby", "F. Scott Fitzgerald", True)
library.add_book("To Kill a Mockingbird", "Harper Lee", True)
library.display_books()

member = Member("John Doe")
library.borrow_book("The Great Gatsby", member)
print(f"{member.name} borrowed 'The Great Gatsby'")
library.display_books()

library.return_book("The Great Gatsby", member)
print(f"{member.name} returned 'The Great Gatsby'")
library.display_books()

library.remove_book("To Kill a Mockingbird")
print("Removed 'To Kill a Mockingbird' from the library")
library.display_books()

# print(library.search_book("The Great Gatsby"))
search_result = library.search_book("The Great Gatsby")
if search_result:
    print(f"Found book: Title: {search_result.title}, Author: {search_result.author}, Available: {'Yes' if search_result.available else 'No'}")
else:
    print("Book not found")

    
```

#### Functional Requirement Question

Design and implement a student management system for a school with the following requirements:

- The system should allow adding new students.
- The system should display all stored students.
- The system should search for a student by a valid identifier or name.
- The system should update existing student details.
- The system should delete a student from the system.
- The system should calculate the average marks of all students.
- The system should count the total number of students in the system.

```python
from dataclasses import dataclass

@dataclass
class Student:
    id: int
    name: str
    age: int
    marks: int

    def display(self):
        print(f"ID: {self.id}, Name: {self.name}, Age: {self.age}, Marks: {self.marks}")


class School:
    def __init__(self):
        self.students = []

    def add_student(self,student:Student):
        self.students.append(student)

    def display_students(self):
        if not self.students:
            print("No students in the school.")
        else:
            print("Students in the school:")
            for student in self.students:
                student.display()

    def search_student(self,student_id:int):
        for student in self.students:
            if student.id == student_id:
                return student
            
        return None
    
    def update_student(self,student_id:int,name:str,age:int,marks:int):
        student = self.search_student(student_id)
        if student:
            student.name = name
            student.age = age
            student.marks = marks
            return True
        return False

    def remove_student(self,student_id:int):
        student = self.search_student(student_id)
        if student:
            self.students.remove(student)
            return True
        return False

    def average_marks(self):
        total_marks = 0
        for student in self.students:
            total_marks += student.marks

        return total_marks / len(self.students)

    def total_students(self):
        return len(self.students)

school = School()

s1 = Student(1,"Alice",15,85)
s2 = Student(2,"Bob",16,90)
school.add_student(s1)
school.add_student(s2)
school.display_students()
school.search_student(1).display()

school.update_student(1,"Alice",15,95)
school.display_students()

school.remove_student(2)
school.display_students()
school.add_student(s2)
print(school.average_marks())
print(school.total_students())
```
