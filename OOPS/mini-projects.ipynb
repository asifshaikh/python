#### Functional Requirement Question

Design and implement a Python library management system that includes:

- a `Book` class with fields:
  - `title`
  - `author`
  - `available`

- a `Member` class with fields:
  - `name`
  - `borrowed_books`

- a `Library` class that uses composition to manage multiple books and provides the following methods:
  - `add_book()`
  - `remove_book()`
  - `search_book()`
  - `borrow_book()`
  - `return_book()`
  - `display_books()`

```python
class Book:
    def __init__(self,title,author,available):
        self.title = title
        self.author = author
        self.available = available

class Member:
    def __init__(self,name):
        self.name = name
        self.borrowed_books = []

class Library:
    def __init__(self,title,author,available):
        self.books = [Book(title,author,available)]

    def add_book(self,title,author,available):
        self.books.append(Book(title,author,available))

    def remove_book(self,title):
        for book in self.books:
            if book.title == title:
                self.books.remove(book)
                return True
        return False

    def search_book(self,title):
        for book in self.books:
            if book.title == title:
                return book
        return None

    def borrow_book(self,title,member):
        book = self.search_book(title)
        if book and book.available:
            book.available = False
            member.borrowed_books.append(book)
            return True
        return False

    def return_book(self,title,member):
        book = self.search_book(title)
        if book and not book.available:
            book.available = True
            member.borrowed_books.remove(book)
            return True
        return False

    def display_books(self):
        print("Available Books:")
        for book in self.books:
            if book.available:
                print(f"Title: {book.title}, Author: {book.author}, Available: Yes")
            

# Usage
library = Library("The Great Gatsby", "F. Scott Fitzgerald", True)
library.add_book("To Kill a Mockingbird", "Harper Lee", True)
library.display_books()

member = Member("John Doe")
library.borrow_book("The Great Gatsby", member)
print(f"{member.name} borrowed 'The Great Gatsby'")
library.display_books()

library.return_book("The Great Gatsby", member)
print(f"{member.name} returned 'The Great Gatsby'")
library.display_books()

library.remove_book("To Kill a Mockingbird")
print("Removed 'To Kill a Mockingbird' from the library")
library.display_books()

# print(library.search_book("The Great Gatsby"))
search_result = library.search_book("The Great Gatsby")
if search_result:
    print(f"Found book: Title: {search_result.title}, Author: {search_result.author}, Available: {'Yes' if search_result.available else 'No'}")
else:
    print("Book not found")

    
```
