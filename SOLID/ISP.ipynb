Interface Segregation Principle:
- It ensures that classes are not forced to implement methods they dont need
- ST:
    - Break large general-purpose interfaces into smaller more specific one's
- ISP is about interfaces

```python
# Bad Example
from abc import ABC,abstractmethod
class Employee(ABC):
    @abstractmethod
    def eat(self):
        pass

    @abstractmethod
    def work(self):
        pass

class Worker(Employee):
    def eat(self):
        print("Eating")

    def work(self):
        print("Working")

class Robot(Employee):
    def work(self):
        print("Working")

    def eat(self):
        raise Exception("Robots cannot eat")

worker = Worker()
worker.eat()
worker.work()

robot = Robot()
robot.work()
# robot.eat()
```

```python
# Good Example
from abc import ABC,abstractmethod

class Workable(ABC):
    @abstractmethod
    def work(self):
        pass

class Eatable(ABC):
    @abstractmethod
    def eat(self):
        pass

class Worker(Workable, Eatable):
    def eat(self):
        print("Eating")

    def work(self):
        print("Working")

class Robot(Workable):
    def work(self):
        print("Working")

worker = Worker()
worker.eat()
worker.work()

robot = Robot()
robot.work()
```
