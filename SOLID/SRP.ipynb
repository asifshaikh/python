Single Responsibility Principle:
- A class should have only one reason to change, meaning a class should do only one job or have only one responsibility
- Dont make one class do everything
- Ex: A User class should only handle user-related data, while db operations should be handled by another class UserRepository

```python
# Bad Example
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age  

    def display(self):
        print(f"Name: {self.name}, Age: {self.age}")

    def add_to_database(self):
        print(f"Adding {self.name} to the database...")

    def remove_from_database(self):
        print(f"Removing {self.name} from the database...")


person = Person("John Doe", 30)
person.display()
person.add_to_database()
person.remove_from_database()
```

```python
# Good Example
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age  

    def display(self):
        print(f"Name: {self.name}, Age: {self.age}")

class PersonRepository:
    def add_to_database(self, person):
        print(f"Adding {person.name} to the database...")

    def remove_from_database(self, person):
        print(f"Removing {person.name} from the database...")

person = Person("John Doe", 30)
person.display()
PersonRepository().add_to_database(person)
PersonRepository().remove_from_database(person)
```
