Dependency Inversion Principle:
- High Level Modules(business logic) should not directly depend on Low Level Module(specific implementation)
- Both should use abstraction
- ST:
    - High Level : Main business logic (what your app does)
    - Low level : specific implementation (db, email etc)
    - Abstraction : interface/contract between them


```python
# Bad Example

class SMSNotification:
    def send(self, message):
        print("Sending SMS:", message)

class EmailNotification:
    def send(self, message):
        print("Sending Email:", message)

class NotificationService:
    def __init__(self,):
        self.email_service = EmailNotification()
        self.sms_service = SMSNotification()

    def send_sms(self,message):
        self.sms_service.send(message)

    def send_email(self,message):
        self.email_service.send(message)


notification_service = NotificationService()
notification_service.send_sms("Hello, World!")
notification_service.send_email("Hello, World!")
```

```python
# Good Example
from abc import ABC, abstractmethod

class NotificationChannel(ABC):
    @abstractmethod
    def send(self, message):
        pass

class SMSService(NotificationChannel):
    def send(self, message):
        print("Sending SMS:", message)

class EmailService(NotificationChannel):
    def send(self, message):
        print("Sending Email:", message)

class NotificationService:
    def __init__(self, notification_channel):
        self.notification_channel = notification_channel

    def send_notification(self, message):
        self.notification_channel.send(message)


notification_channel = SMSService()
notification_service = NotificationService(notification_channel)
notification_service.send_notification("Hello, World! from SMS")

notification_channel = EmailService()
notification_service = NotificationService(notification_channel)
notification_service.send_notification("Hello, World! from Email")
```
