```python
# Balanced Parentheses
# Explain solution: The function `is_balanced` checks if a string of parentheses is balanced. It uses a stack to keep track of opening brackets. For each character in the string, if it's an opening bracket, it gets pushed onto the stack. If it's a closing bracket, the function checks if the stack is empty or if the top of the stack does not match the corresponding opening bracket. If either condition is true, the string is not balanced, and the function returns False. If the loop completes and the stack is empty, it means all brackets were matched correctly, and the function returns True.
def is_balanced(s):
    stack = []
    mapping = {')': '(', '}': '{', ']': '['}

    for char in s:
        if char in mapping.values():
            stack.append(char)
        elif char in mapping.keys():
            if not stack or mapping[char] != stack.pop():
                return False
    return not stack

# Test cases
print(is_balanced("()"))  # True
print(is_balanced("()[]{}"))  # True
print(is_balanced("(]"))  # False
print(is_balanced("{[()]}"))  # True

# dry run for "([)]"
# 1. '(' is an opening bracket, so push it onto the stack. Stack: ['(']
# 2. '[' is an opening bracket, so push it onto the stack. Stack: ['(', '[']
# 3. ')' is a closing bracket. The top of the stack is '[', which does not match ')'. Return False.
```

```python
# postfix evaluation

def evaluate_postfix(expression):
    stack = []
    for token in expression.split():
        if token.isdigit():
            stack.append(int(token))
        else:
            right = stack.pop()
            left = stack.pop()
            if token == '+':
                stack.append(left + right)
            elif token == '-':
                stack.append(left - right)
            elif token == '*':
                stack.append(left * right)
            elif token == '/':
                stack.append(left / right)
    return stack.pop()

# Test cases
print(evaluate_postfix("2 3 +"))  # 5
print(evaluate_postfix("5 1 2 + 4 * + 3 -"))  # 14
```

```python
# prefix evaluation

def evaluate_prefix(expression):
    stack = []
    for token in reversed(expression.split()):
        if token.isdigit():
            stack.append(int(token))
        else:
            right = stack.pop()
            left = stack.pop()
            if token == '+':
                stack.append(left + right)
            elif token == '-':
                stack.append(left - right)
            elif token == '*':
                stack.append(left * right)
            elif token == '/':
                stack.append(left / right)
    return stack.pop()

# Test cases
print(evaluate_prefix("+ 2 3"))  # 5
```

```python
# Text Editor

class TextEditor:
    def __init__(self):
        self.text = []
        self.cursor = 0

    def insert(self, char):
        self.text.insert(self.cursor, char)
        self.cursor += 1

    def delete(self):
        if self.cursor > 0:
            self.text.pop(self.cursor - 1)
            self.cursor -= 1

    def display(self):
        return ''.join(self.text)

# Test cases
editor = TextEditor()
editor.insert('Hello')
editor.insert(' ')
editor.insert('World')
print(editor.display())  # Hello World
```

```python
# Min Stack

class MinStack:
    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, val):
        self.stack.append(val)
        if not self.min_stack or val <= self.min_stack[-1]:
            self.min_stack.append(val)

    def pop(self):
        if self.stack.pop() == self.min_stack[-1]:
            self.min_stack.pop()

    def top(self):
        return self.stack[-1]

    def getMin(self):
        return self.min_stack[-1]

# Test cases
min_stack = MinStack()
min_stack.push(-2)
min_stack.push(0)
min_stack.push(-3)
print(min_stack.getMin())  # -3
```
