1. What is a Linked list?
- Linked list is a non linear data structure. Instead of storing data everything together it stores each element separately and then connect them.

2. What is a Node?
- A linked list is mode of Nodes
- A Node contains two things:
    - data 
    - next node reference

```python
# HEAD
#  ↓
# +----+------+     +----+------+     +----+------+
# |10  |  •-------> |20  |  •-------> |30  | None |
# +----+------+     +----+------+     +----+------+
```

3. What is Head?
- The linked list itself only remembers one thing: and ie Head
- If we know the head we can reach every node, without the head the list is lost

4. Creating our first Node Class

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
```

5. Creating Nodes

```python
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
```

6. Connecting Nodes

```python
node1.next = node2
node2.next = node3
```

7. Creating the Head

```python
head = node1
```

8. Accessing Data

```python
print(head.data)
print(head.next.data)
print(head.next.next.data)
```

9. Traversing a Linkedlist

```python
current = head
while current is not None:
    print(current.data)
    current = current.next
```

10. Creating a Linkedlist class

```python
class LinkedList:
    def __init__(self):
        self.head = Node
    
    def isEmpty(self):
        return self.head is None
    
    def display(self):
        current = self.head
        while current is not None:
            print(current.data, end=" -> ")
            current = current.next
        
        print("None")
    
    def length(self):
        current = self.head
        count = 0
        while current is not None:
            count += 1
            current = current.next
        
        return count
    
    def search(self,target):
        current = self.head
        while current is not None:
            if current.data == target:
                return True
            current = current.next
        
        return False

```

```python
l1 = LinkedList()
l1.head = node1
l1.display()
l1.length()

```
