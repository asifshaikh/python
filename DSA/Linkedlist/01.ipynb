1. What is a Linked list?
- Linked list is a non linear data structure. Instead of storing data everything together it stores each element separately and then connect them.

2. What is a Node?
- A linked list is mode of Nodes
- A Node contains two things:
    - data 
    - next node reference

```python
# HEAD
#  ↓
# +----+------+     +----+------+     +----+------+
# |10  |  •-------> |20  |  •-------> |30  | None |
# +----+------+     +----+------+     +----+------+
```

3. What is Head?
- The linked list itself only remembers one thing: and ie Head
- If we know the head we can reach every node, without the head the list is lost

4. Creating our first Node Class

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
```

5. Creating Nodes

```python
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
```

6. Connecting Nodes

```python
node1.next = node2
node2.next = node3
```

7. Creating the Head

```python
head = node1
```

8. Accessing Data

```python
print(head.data)
print(head.next.data)
print(head.next.next.data)
```

9. Traversing a Linkedlist

```python
current = head
while current is not None:
    print(current.data)
    current = current.next
```

10. Creating a Linkedlist class

```python
class LinkedList:
    def __init__(self):
        self.head = Node
    
    def isEmpty(self):
        return self.head is None
    
    def display(self):
        current = self.head
        while current is not None:
            print(current.data, end=" -> ")
            current = current.next
        
        print("None")
    
    def length(self):
        current = self.head
        count = 0
        while current is not None:
            count += 1
            current = current.next
        
        return count
    
    def search(self,target):
        current = self.head
        while current is not None:
            if current.data == target:
                return True
            current = current.next
        
        return False

```

```python
l1 = LinkedList()
l1.head = node1
l1.display()
l1.length()

```

11. Types of Insertion
- Insertion at **Beginning** -> its O(1)
- Insertion at **End**
- Insertion at **Position**

```python
def insert_at_beginning(self,data):
    new_node = Node(data)
    new_node.next = self.head
    self.head = new_node

def insert_at_end(self,data):
    new_node = Node(data)

    if self.head is None:
        self.head = new_node
        return

    current = self.head
    while current.next is not None:
        current = current.next

    current.next = new_node

def insert_at_position(self,position,data):
    if position == 0:
        self.insert_at_beginning(data)
        return

    new_node = Node(data)

    current = self.head
    for _ in range(position - 1):
        if current is None:
            raise IndexError("Position out of range")

        current = current.next

    if current is None:
        raise IndexError("Position out of range")

    new_node.next = current.next
    current.next = new_node
```

13. Deletion in Linkedlist
- before:
    - 10->20->30
- after:
    - 10 --->30
- we effectively bypass the node
- Python's garbage collector can eventually reclaim the object if nothing else refers to it

```python
# delete from beginning
# HEAD
#  ↓
# 10 → 20 → 30 → None
# HEAD
#  ↓
# 20 → 30 → None
def delete_from_beginning(self):
    if self.head is None:
        raise IndexError("cannot delete from an empty list")

    self.head = self.head.next
```

```python
# Delete from end
# HEAD
#  ↓
# 10 → 20 → 30 → 40 → None 
# we want to delete 40
def delete_from_end(self):
    if self.head is None:
        raise IndexError("Cannot delete from an empty list")

    if self.head.next is None:
        self.head = None
        return
    current = self.head
    while current.next.next is not None:
        current = current.next

    current.next = None
```

```python
# Delete by value
def delete_by_value(self,value):
    if self.head is None:
        raise IndexError("Cannot delete from an empty list")

    if self.head.data == value:
        self.head = self.head.next
        return

    current = self.head
    while current.next is not None:
        if current.next.data == value:
            current.next = current.next.next
            return

        current = current.next

    raise ValueError("Value not found")
```

```python
# delete by position
def delete_at_position(self,position):
    if position < 0:
        raise IndexError("Position cannot be negative")

    if self.head is None:
        raise IndexError("Cannot delete item from empty list")

    if position == 0:
        self.head = self.head.next
        return

    current = self.head
    for _ in range(position - 1):
        if current.next is None:
            raise IndexError("Position out of range")

        current = current.next

    if current.next is None:
        raise IndexError("Position out of range")

    current.next = current.next.next
```

```python
# delete all occurences of a particular value
def delete_all(self,value):
    while self.head is not None and self.head.data == value:
        self.head = self.head.next

    current = self.head

    while current is not None and current.next is not None:
        if current.next.data == value:
            current.next = current.next.next
        else:
            current = current.next
```

14. Complete Linked list code including Insertion and Deletion

```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:

    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def display(self):

        current = self.head

        while current is not None:
            print(current.data, end=" -> ")
            current = current.next

        print("None")

    def length(self):

        count = 0
        current = self.head

        while current is not None:
            count += 1
            current = current.next

        return count

    def search(self, target):

        current = self.head

        while current is not None:

            if current.data == target:
                return True

            current = current.next

        return False

    # -------------------------
    # INSERTION
    # -------------------------

    def insert_at_beginning(self, data):

        new_node = Node(data)

        new_node.next = self.head
        self.head = new_node

    def insert_at_end(self, data):

        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head

        while current.next is not None:
            current = current.next

        current.next = new_node

    def insert_at_position(self, position, data):

        if position < 0:
            raise IndexError("Position cannot be negative")

        if position == 0:
            self.insert_at_beginning(data)
            return

        new_node = Node(data)

        current = self.head

        for _ in range(position - 1):

            if current is None:
                raise IndexError("Position out of range")

            current = current.next

        if current is None:
            raise IndexError("Position out of range")

        new_node.next = current.next
        current.next = new_node

    # -------------------------
    # DELETION
    # -------------------------

    def delete_from_beginning(self):

        if self.head is None:
            raise IndexError("Cannot delete from an empty list")

        self.head = self.head.next

    def delete_from_end(self):

        if self.head is None:
            raise IndexError("Cannot delete from an empty list")

        if self.head.next is None:
            self.head = None
            return

        current = self.head

        while current.next.next is not None:
            current = current.next

        current.next = None

    def delete_by_value(self, value):

        if self.head is None:
            raise ValueError("Cannot delete from an empty list")

        if self.head.data == value:
            self.head = self.head.next
            return

        current = self.head

        while current.next is not None:

            if current.next.data == value:
                current.next = current.next.next
                return

            current = current.next

        raise ValueError("Value not found")

    def delete_at_position(self, position):

        if position < 0:
            raise IndexError("Position cannot be negative")

        if self.head is None:
            raise IndexError("Cannot delete from an empty list")

        if position == 0:
            self.head = self.head.next
            return

        current = self.head

        for _ in range(position - 1):

            if current.next is None:
                raise IndexError("Position out of range")

            current = current.next

        if current.next is None:
            raise IndexError("Position out of range")

        current.next = current.next.next
```
