1) What is Doubly Linkedlist?
- A Singly Linked List knows where to go next. A Doubly Linked List knows where to go next AND where it came from.

Node: 
- Previous <- Data -> Next

```python
class Node:
    def __init__(self,data):
        self.prev = None
        self.data = data
        self.next = None

node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.prev = node1
node2.next = node3
node3.prev = node2
```

```python
class DoublyLinkedList:

    def __init__(self):
        self.head = None
        self.tail = None
```

```python
# Insert at beginning
# HEAD
#  ↓
# 10 ⇄ 20 ⇄ 30
# HEAD
#  ↓
# 5 ⇄ 10 ⇄ 20 ⇄ 30
def insert_at_beginning(self,data):
    new_node = Node(data)
    if self.head is None:
        self.head = new_node
        self.tail = new_node
        return
    
    new_node.next = self.head
    self.head.prev = new_node
    self.head = new_node
    
```

```python
# Insert at end
# HEAD
#  ↓
# 10 ⇄ 20 ⇄ 30
#           ↑
#          TAIL

def insert_at_end(self,data):
    new_node = Node(data)
    if self.tail is None:
        self.head = new_node
        self.tail = new_node
        return
    new_node.prev = self.tail
    self.tail.next = new_node
    self.tail = new_node
    
```

```python
# display forward
def display_forward(self):
    current = self.head
    while current:
        print(current.data, end=" ")
        current = current.next
    print("None")
```

```python
# display backward
def display_backward(self):

    current = self.tail

    while current is not None:
        print(current.data, end=" ⇄ ")
        current = current.prev

    print("None")
```

```python
# Insert at position
# 20 ⇄ 30
# 20 ⇄ 25 ⇄ 30
# 25.prev = 20
# 25.next = 30

# 20.next = 25
# 30.prev = 25
def insert_at_position(self, position, data):

    if position < 0:
        raise IndexError("Position cannot be negative")

    if position == 0:
        self.insert_at_beginning(data)
        return

    if self.head is None:
        raise IndexError("Position out of range")

    current = self.head

    for _ in range(position - 1):

        if current.next is None:
            raise IndexError("Position out of range")

        current = current.next

    if current.next is None:
        self.insert_at_end(data)
        return

    new_node = Node(data)

    new_node.next = current.next
    new_node.prev = current

    current.next.prev = new_node
    current.next = new_node
```

```python
# delete from beginning

def delete_from_beginning(self):
    if self.head is None:
        return

    if self.head.next is None:
        self.head = None
        self.tail = None
        return

    self.head = self.head.next
    self.head.prev = None
```

```python
# delete from end
def delete_from_end(self):

    if self.tail is None:
        raise IndexError("Cannot delete from an empty list")

    if self.head is self.tail:
        self.head = None
        self.tail = None
        return

    self.tail = self.tail.prev
    self.tail.next = None
```
