1) What is Doubly Linkedlist?
- A Singly Linked List knows where to go next. A Doubly Linked List knows where to go next AND where it came from.

Node: 
- Previous <- Data -> Next

```python
class Node:
    def __init__(self,data):
        self.prev = None
        self.data = data
        self.next = None

node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.prev = node1
node2.next = node3
node3.prev = node2
```
