```python
class Node:

    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:

    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def display(self):

        current = self.head

        while current is not None:
            print(current.data, end=" -> ")
            current = current.next

        print("None")

    def length(self):

        count = 0
        current = self.head

        while current is not None:
            count += 1
            current = current.next

        return count

    def search(self, target):

        current = self.head

        while current is not None:

            if current.data == target:
                return True

            current = current.next

        return False

    # -------------------------
    # INSERTION
    # -------------------------

    def insert_at_beginning(self, data):

        new_node = Node(data)

        new_node.next = self.head
        self.head = new_node

    def insert_at_end(self, data):

        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        current = self.head

        while current.next is not None:
            current = current.next

        current.next = new_node

    def insert_at_position(self, position, data):

        if position < 0:
            raise IndexError("Position cannot be negative")

        if position == 0:
            self.insert_at_beginning(data)
            return

        new_node = Node(data)

        current = self.head

        for _ in range(position - 1):

            if current is None:
                raise IndexError("Position out of range")

            current = current.next

        if current is None:
            raise IndexError("Position out of range")

        new_node.next = current.next
        current.next = new_node

    # -------------------------
    # DELETION
    # -------------------------

    def delete_from_beginning(self):

        if self.head is None:
            raise IndexError("Cannot delete from an empty list")

        self.head = self.head.next

    def delete_from_end(self):

        if self.head is None:
            raise IndexError("Cannot delete from an empty list")

        if self.head.next is None:
            self.head = None
            return

        current = self.head

        while current.next.next is not None:
            current = current.next

        current.next = None

    def delete_by_value(self, value):

        if self.head is None:
            raise ValueError("Cannot delete from an empty list")

        if self.head.data == value:
            self.head = self.head.next
            return

        current = self.head

        while current.next is not None:

            if current.next.data == value:
                current.next = current.next.next
                return

            current = current.next

        raise ValueError("Value not found")

    def delete_at_position(self, position):

        if position < 0:
            raise IndexError("Position cannot be negative")

        if self.head is None:
            raise IndexError("Cannot delete from an empty list")

        if position == 0:
            self.head = self.head.next
            return

        current = self.head

        for _ in range(position - 1):

            if current.next is None:
                raise IndexError("Position out of range")

            current = current.next

        if current.next is None:
            raise IndexError("Position out of range")

        current.next = current.next.next
```

2. Three Pointer Technique
- previous
- current
- next_node

- Initially:
    - previous = None
    - current = head

- Every Iteration follows the same four steps
1. SAVE \
   next_node = current.next

2. REVERSE \
   current.next = previous

3. MOVE previous \
   previous = current

4. MOVE current \
   current = next_node

```python
def reverse_linked_list(self):
    previous = None
    current = self.head

    while current is not None:
        new_node = current.next
        current.next = previous
        previous = current
        current = new_node

    self.head = previous
```

2. Slow and Fast Pointers
- Use two references:
    - slow -> moves 1 step
    - fast -> moves 2 steps

```python
# Find middle of linkedlist
def find_middle(head):
    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

    return slow

# usage

head = Node(1)
head.next = Node(2)
head.next.next = Node(3)
head.next.next.next = Node(4)
head.next.next.next.next = Node(5)

middle_node = find_middle(head)
print(middle_node.data)
```

```python
# NTH NODE FROM THE END
# slow and fast pointer
# we move fast by n times and then slow will be at nth node from the end
def nth_node_from_end(head, n):
    slow = head
    fast = head
    for _ in range(n):
        if fast is None:
            return None
        fast = fast.next

    while fast is not None:
        slow = slow.next
        fast = fast.next

    return slow

nth_node = nth_node_from_end(head, 1)
print(nth_node.data)
```

```python
# Detect Cycle in Linkedlist
def detect_cycle(head):
    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

        if slow is fast:
            return True

    return False

# usage

head = Node(1)
head.next = Node(2)
head.next.next = Node(3)
head.next.next.next = Node(4)
head.next.next.next.next = Node(5)
head.next.next.next.next.next = head.next

has_cycle = detect_cycle(head)
print(has_cycle)
```

```python
# Finding the start of the cycle
def find_cycle_start(head):
    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

        if slow is fast:
            break

    pointer1 = head
    pointer2 = slow

    while pointer1 is not pointer2:
        pointer1 = pointer1.next
        pointer2 = pointer2.next

    return pointer1

# usage

head = Node(1)
head.next = Node(2)
head.next.next = Node(3)
head.next.next.next = Node(4)
head.next.next.next.next = Node(5)
head.next.next.next.next.next = head.next.next

start_node = find_cycle_start(head)
print(start_node.data)
```

```python
# remove cycle
def remove_cycle(head):
    cycle_start = find_cycle_start(head)
    current = cycle_start

    while current.next is not cycle_start:
        current = current.next

    current.next = None

# usage

head = Node(1)
head.next = Node(2)
head.next.next = Node(3)
head.next.next.next = Node(4)
head.next.next.next.next = Node(5)
head.next.next.next.next.next = head.next.next

remove_cycle(head)
```
