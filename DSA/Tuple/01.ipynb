1) What is Tuple?
- A tuple is an ordered,immutable collection of elements
- A tuple is similar to list but biggest difference is List is mutable whereas tuple is immutable

3) Creating a Tuple

```python
# Using parentheses
a = (1, 2, 3)
print(a)

# Using the tuple() constructor
a = tuple((1, 2, 3))
print(a)

# multiple data types
a = (1, "Hello", 3.4)
print(a)

# empty tuple
a = ()
print(a)

# tuple with one element
a = (5,)
print(a)

# string in tuple
a = tuple("Hello")
print(a)
```

3) Accessing Tuple Elements
- Tuples support indexing exactly like lists.

```python
numbers = (1, 2, 3, 4, 5)

print(numbers[0])  # Output: 1
print(numbers[-1])  # Output: 5
```

4) Tuple Slicing

```python
numbers = (10, 20, 30, 40, 50)

print(numbers[1:4])
print(numbers[::-1])
```

5) Can we Delete a Tuple?
- You cannot delete an individual tuple element
- You can delete the entire tuple variable

```python
numbers = (10, 20, 30)

# del numbers[0] # This will raise a TypeError
del numbers
# print(numbers)
```

6) Tuple Methods
- Unlike lists, tuples have only two primary methods.
- `count()` and `index()`

```python
numbers = (10, 20, 20, 30, 20)

print(numbers.count(20))
print(numbers.index(20))
```

7) Tuple Packing
- Packing means putting multiple values into a tuple.
- `student = "John", 25, 90` python interprets this as: `student = ("John", 25, 90)`

8) Tuple Unpacking

```python
student = ("John", 20, "Computer Science")

name, age, course = student

print(name)
print(age)
print(course)
```

8) Extended Unpacking
- Python gives us `*`.

```python
numbers = (10, 20, 30, 40, 50)

a, *b = numbers

print(a)  # Output: 10
print(b)  # Output: [20, 30, 40, 50]
first, *middle, last = numbers

print(first)  # Output: 10
print(middle)  # Output: [20, 30, 40]
print(last)  # Output: 50
```

```python
x = (1, 2, 3)
print(x)
x += (4, 5)
print(x)
```
