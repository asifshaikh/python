1) What is Tuple?
- A tuple is an ordered,immutable collection of elements
- A tuple is similar to list but biggest difference is List is mutable whereas tuple is immutable


2) List vs Tuple
| Feature              | List | Tuple      |
| -------------------- | ---- | ---------- |
| Ordered              | ✅    | ✅          |
| Indexed              | ✅    | ✅          |
| Slicing              | ✅    | ✅          |
| Duplicates           | ✅    | ✅          |
| Different data types | ✅    | ✅          |
| Mutable              | ✅    | ❌          |
| `append()`           | ✅    | ❌          |
| `remove()`           | ✅    | ❌          |
| `pop()`              | ✅    | ❌          |
| `sort()`             | ✅    | ❌          |
| `count()`            | ✅    | ✅          |
| `index()`            | ✅    | ✅          |
| Hashable             | ❌    | Sometimes* |
