1) What is a Dictionary?
- A dictionary stores data as: key -> value

```python
# Example
student={
    "name":"asif",
    "age":20
}
print(student)

# instead of accesing data using numerical index we can access data using key
print(student["name"])
print(student["age"])
```

2) Properties


| Property                      | Dictionary   |
| ----------------------------- | ------------ |
| Key-value structure           | ✅            |
| Mutable                       | ✅            |
| Duplicate keys                | ❌            |
| Duplicate values              | ✅            |
| Indexed by integer position   | ❌            |
| Keys must be hashable         | ✅            |
| Values can be almost anything | ✅            |
| Fast key lookup               | O(1) average |
| Insertion order preserved     | ✅            |

5) Accessing Values using `[]`

```python
student = {
    "name":"asif",
    "age":20
}

print(student["name"])
print(student["age"])
```

6) If a key doesnt exists it will raise a KeyError

7) `.get()`

```python
student = {
    "name":"asif",
    "age":20
}
print(student.get("name"))
print(student.get("age"))
print(student.get("address")) # Will return None as address is not present instead of raising KeyError
```

8) `[]` vs `.get()`

| `dict[key]`                  | `dict.get(key)`           |
| ---------------------------- | ------------------------- |
| Raises `KeyError` if missing | Returns `None` by default |
| Useful when key must exist   | Useful for optional keys  |
| Direct lookup                | Safe lookup               |


9) `.get()` with default value

```python
salary = student.get("salary", 0)
print(salary)
```

10) Adding a new Key
- simply assign a value

```python
student["city"] = "Goa"
print(student)
```

11) Update an Existing Key

```python
student["city"] = "Vasco"
print(student)
```

12. Duplicate Keys

```python
student = {
    "name": "asif",
    "age": 20,
    "age":23
}
print(student) # The last value will be considered
```

13. `.update()`
- can add or modify multiple key-value pairs

```python
student = {
    "name":"asif",
    "age":20
}
student.update({
    "age":23,
    "city":"Goa"
})
print(student)
```

14. `pop()`
- removes a key-value pair and returns the value

```python
student = {
    "name": "asif",
    "age": 20
}
age = student.pop("age")
print(age)
print(student)
```

15. `popitem()`
- removes and returns the last inserted key-value pair

```python
student = {
    "name": "asif",
    "age": 20,
    "city": "Goa"
}

item = student.popitem()
print(item)
print(student)

```

16. `del`
- we can also delete a key using del

```python
student = {
    "name": "asif",
    "age": 20,
    "city": "Goa"
}

del student["age"]
print(student)
```

17. `clear()`
- removes all entries

```python
student = {
    "name": "asif",
    "age": 20,
    "city": "Goa"
}

student.clear()
print(student)
```

18. `keys()`
- returns a view of dictionary's keys

```python
student = {
    "name": "asif",
    "age": 20,
    "city": "Goa"
}
print(student.keys())
for key in student.keys():
    print(key)

```

19. `values()`
- returns a view of all values

```python
print(student.values())
for value in student.values():
    print(value)
```

20. `items()`
- returns key-value pairs.

```python
for key, value in student.items():
    print(key, value)
```

```python
def two_sum(nums:list[int],target:int) -> list[int]:
    num_map = {}
    for i in range(len(nums)):
        diff = target - nums[i]
        if diff in num_map:
            return [num_map[diff],i]
        num_map[nums[i]] = i
    return None

nums = [2,7,11,15]
target = 9
print(two_sum(nums,target))
```

```python
# Grouping Data
students = [
    ("asif","A"),
    ("mohsin","B"),
    ("zain","A")
]

student_dict = {}
for name, grade in students:
    if grade not in student_dict:
        student_dict[grade] = []
    student_dict[grade].append(name)

print(student_dict)
```
