1) What is a Dictionary?
- A dictionary stores data as: key -> value

```python
# Example
student={
    "name":"asif",
    "age":20
}
print(student)

# instead of accesing data using numerical index we can access data using key
print(student["name"])
print(student["age"])
```

2) Properties


| Property                      | Dictionary   |
| ----------------------------- | ------------ |
| Key-value structure           | ✅            |
| Mutable                       | ✅            |
| Duplicate keys                | ❌            |
| Duplicate values              | ✅            |
| Indexed by integer position   | ❌            |
| Keys must be hashable         | ✅            |
| Values can be almost anything | ✅            |
| Fast key lookup               | O(1) average |
| Insertion order preserved     | ✅            |

5) Accessing Values using `[]`

```python
student = {
    "name":"asif",
    "age":20
}

print(student["name"])
print(student["age"])
```

6) If a key doesnt exists it will raise a KeyError

7) `.get()`

```python
student = {
    "name":"asif",
    "age":20
}
print(student.get("name"))
print(student.get("age"))
print(student.get("address")) # Will return None as address is not present instead of raising KeyError
```

8) `[]` vs `.get()`

| `dict[key]`                  | `dict.get(key)`           |
| ---------------------------- | ------------------------- |
| Raises `KeyError` if missing | Returns `None` by default |
| Useful when key must exist   | Useful for optional keys  |
| Direct lookup                | Safe lookup               |


9) `.get()` with default value

```python
salary = student.get("salary", 0)
print(salary)
```

10) Adding a new Key
- simply assign a value

```python
student["city"] = "Goa"
print(student)
```

11) Update an Existing Key

```python
student["city"] = "Vasco"
print(student)
```
