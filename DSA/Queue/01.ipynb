1) What is Queue?
- Its a linear data structure that follows the FIFO principle
- First In First Out
- The element that enters the queue first is the element that leaves first

2) Real World Examples of Queue
- Bank Queue
    - Customer A -> Customer B -> Customer C
    - A arrives first -> gets served first

- Printer Queue
    - Document 1 -> Document 2 -> Document 3

- CPU Scheduling
    - Process p1 -> Process p2 -> Process p3

3) Basic Queue Terminology
- A queue has two important ends:

              Queue
        ┌────┬────┬────┬────┐
        │ A  │ B  │ C  │ D  │
        └────┴────┴────┴────┘
         ↑              ↑
       FRONT           REAR

4) Implement Queue using Python List

```python
queue = []

queue.append(10)
queue.append(20)
queue.append(30)

print(queue)

queue.pop(0) # O(N) because it has to shift all the elements to the left
print(queue)
```

5) Implement Queue using `collections.deque`

```python
from collections import deque

queue = deque()

queue.append(10)
queue.append(20)
queue.append(30)

print(list(queue))

queue.popleft() # O(1) because it doesn't have to shift all the elements to the left
print(queue)
```

```python
# important deque methods
from collections import deque

queue = deque()

queue.append(10) # adds to the right/rear
queue.appendleft(20) # adds to the left/front
queue.pop() # removes from the right/rear
queue.popleft() # removes from the left/front
queue.extend([30, 40, 50]) # adds multiple elements to the right/rear
queue.extendleft([60, 70, 80]) # adds multiple elements to the left/front
queue.rotate(3) # rotates the deque n steps to the right. If n is negative, rotates to the left. 
print(queue)
```
