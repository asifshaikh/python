1) What is a Set?
- A set is an **unordered collection of unique elements**

2) Properties of Set 
| Property             | Set            |
| -------------------- | -------------- |
| Ordered              | ❌              |
| Indexed              | ❌              |
| Mutable              | ✅              |
| Duplicates           | ❌              |
| Different data types | ✅              |
| Slicing              | ❌              |
| Membership testing   | Very efficient |
| Hash-based           | ✅              |


3) Creating a Set
- using `{}`
- `numbers = {}` -> this creates a dictionary not a set


```python
numbers = {}
print(type(numbers))
numbers = set()
print(type(numbers))
```

4) Creating a Set from a List

```python
numbers = [10,20,30,40,50,50,40,30,20,10]

unique_nums = set(numbers)
print(unique_nums)
```

5) Creating a Set from a String

```python
letters = set('banana')
print(letters)
```

6) Set doesnt have indexing like list
- Why? Because a set doesn't maintain the positional sequence that a list does.

```python
numbers = {10,20,30,40,50}
print(numbers[0])
```

7) `add()`
- adds one element to a set

```python
numbers = {10,20,30,40,50}
numbers.add(60)
print(numbers)

numbers.add(40)
print(numbers)
```

8) `update()`
- adds multiple elements from an iterable

```python
numbers = {10,20,30,40,50}

numbers.update([60,70,80])
print(numbers)

numbers.update([40,50,60])
print(numbers)
```

9) `remove()`
- removes an element

```python
numbers = {10,20,30,40,50}

numbers.remove(40)
print(numbers)
```

10) `discard()`
- also removes element, but if the element doesnt exist nothing happens, whereas `remove()` raises an error 

```python
numbers = {10,20,30,40,50}

numbers.discard(40)
print(numbers)

numbers.discard(100)
print(numbers)
```

11) `pop()`
- A set also has `pop()`
- there is difference between from list `pop()`
- with a list: removes the last element
- with a set: removes an arbitrary element since sets dont have indexes

12) `clear()`
- removes everything


```python
numbers = {10,20,30,40,50}
numbers.clear()
print(numbers)
```

13) Membership Testing
- `O(1)` generally for set
- `O(n)` for list

```python
numbers = {10,20,30,40,50}

print(10 in numbers)
print(100 in numbers)
```

14) Set Operations
- `union()` -> adds all elements from both sets
- `intersection()` -> elements present in both sets
- `difference()` -> elements that are in A but not in B
- `symmetric_difference()` -> elements that are in either set, but not in both
- `issubset()` -> a set A is a subset of B if every element of A exists in B.
- `issuperset()` -> a set is a superset if it contains all elements of another set

```python
a = {1,2,3}
b = {3,4,5}

print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
print(b.difference(a))
print(a.symmetric_difference(b))

c = {1,2}
d = {1,2,3,4}

print(c.issubset(d))
print(d.issubset(c))

print(c.issuperset(d))
print(d.issuperset(c))
```
