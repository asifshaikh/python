1. What is a String?
- its a sequence of characters
- it supports : 
    - indexing
    - slicing
    - iteration
    - membership testing

```python
name = "Python"
print(name)
print(type(name))
```

2. Creating strings

```python
# single quotes
name = 'Python'

# double quotes
name = "Python"

# triple quotes
name = """This is a 
multi-line string."""

print(name)
print(type(name))
```

3. String Indexing

```python
text = 'Python'
print(text[0])
print(text[1])
print(text[2])
print(text[3])
print(text[4])
print(text[5])
print(text[-1]) # Negative indexing
print(text[-2])
print(text[-3])
print(text[-4])
print(text[-5])
print(text[-6])
```

4. String Slicing
- syntax string[start:stop:step]

```python
text = 'Python'

print(text[0:2])
print(text[2:5])
print(text[5:])
print(text[::-1])
```

5. Strings are immutable

```python
text = 'Python'

text[0] = "j"
```

6. How do we modify a string?
- we create a new string

```python
text = 'python'

text = 'j' + text[1:]
print(text)
```

7. `len()`
- get the number of characters

```python
text = 'python'
print(len(text))
```

8. Traversing a string

```python
text = 'python'

for char in text:
    print(char)
```

9. Membership Testing

```python
if "p" in text:
    print("yes")
```

10. `lower()`
- converts characters to lowercase

```python
text = "PYTHON"

print(text.lower())
print(text) # original string is not changed, since string is immutable
```

11. `upper()`
- converts characters to uppercase

```python
text = "python"

print(text.upper())
print(text)
```

12. `capitalize()`
- only the first letter is capitalized

```python
text = "python programming"

print(text.capitalize())
print(text)
```

13. `title()`
- first letter of each word is capitalized

```python
text = "python programming"

print(text.title())
print(text)
```

14. `strip()`
- removes whitespace from beginning and end

```python
text = "   Python   "

print(text.strip())
print(text)
```

15. `lstrip()` adn `rstrip()`
- `lstrip()` removes whitespace from the left
- `rstrip()` removes whitespace from the right

16. `replace()`
- replace one substring with another

```python
text = " I like Java "
text = text.replace("Java", "Python")
print(text)
```

17. `split()`

```python
text = " I like Java "
word = text.split()
print(word)
```

18. `split()` with a seperator

```python
data = "apple,banana,orange"

fruits = data.split(",")
print(fruits) # returns a list
```

```python
date = "16-08-2026"
parts = date.split("-")
print(parts)
```

19. `join()`

```python
words = ["Python","is","a","programming","language"]

sentence = " ".join(words)
print(sentence)
```

20. `find()`
- find the first occurence of a substring

```python
text = "hello world"
print(text.find("world"))
```

21. `index()`

```python
text = "hello world"
print(text.index("world"))
# print(text.index("python")) # throws an error
```

22. `count()`
- counts occurences

```python
text = "banana"
print(text.count("a"))
print(text.count("an"))
```

23. `startswith()` and `endswith()`

```python
text = "hello world"
print(text.startswith("hello"))
print(text.startswith("Hello"))
print(text.endswith("world"))
print(text.endswith("World"))
```

24. Character Classification Methods

```python
# isalpha() -> returns True if all characters in the string are alphabetic
print("Python".isalpha())
print("Python123".isalpha())
```

```python
# isdigit() -> returns True if all characters in the string are digits
print("123".isdigit())
print("Python123".isdigit())
```

```python
# isalnum() -> returns True if all characters in the string are alphanumeric
print("123".isalnum())
print("Python123".isalnum())
print("Python".isalnum())
```

```python
# isspace() -> returns True if all characters in the string are spaces
print("   ".isspace())
print("   Python   ".isspace())
print("Python".isspace())
```

```python
# islower() -> returns True if all characters in the string are lowercase
print("python".islower())
print("Python".islower())
print("PYTHON".islower())
```

```python
# isupper() -> returns True if all characters in the string are uppercase
print("python".isupper())
print("Python".isupper())
print("PYTHON".isupper())
```

26. String Formatting
- f-strings
- string concatenation

```python
name = "asif"
age = 23
print(f"Name: {name}, Age: {age}")
```

```python
first = "hello"
second = "world"

result = first + " " + second
print(result)
print(f"{first} {second}")
```
