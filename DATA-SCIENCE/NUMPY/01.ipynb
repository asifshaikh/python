```python
import numpy as np

print(np.__version__)
```

1. Creating arrays

```python
list1 = [1,2,3]
print(list1)
```

```python
arr_list1 = np.array(list1) # Creating array using list
print(arr_list1)
```

```python
print(type(arr_list1)) # check data type of variable
```

```python
print(arr_list1.ndim) # check dimension of array
```

```python
# Creating 2D array using 2 lists
list1 = [1,2,3]
list2 = [4,5,6]
arr_list2 = np.array([list1,list2])
print(arr_list2)
```

```python
print(arr_list2.ndim)
```

```python
# Creating 3D array using 3 lists
list1 = [1,2,3]
list2 = [4,5,6]
list3 = [7,8,9]
arr_list3 = np.array([[list1,list2,list3]])
print(arr_list3)
```

```python
print(arr_list3.ndim)
```

2. numpy array attributes

```python
arr1 = np.array([[1,2,3],[4,5,6]])
print(arr1)
```

```python
# Attributes of Array (properties of array)
print("Shape", arr1.shape) # shape = rows X columns
print("Size", arr1.size) # number of elements in array
print("Data type", arr1.dtype) # data type of array
print("Dimension", arr1.ndim) # dimension of array
```

```python
# 3D array
list1 = [1,2,3]
list2 = [4,5,6]
list3 = [7,8,9]
arr2 = np.array([[list1,list2,list3]])

print("Shape", arr2.shape) # shape = rows X columns
print("Size", arr2.size) # number of elements in array
print("Data type", arr2.dtype) # data type of array
print("Dimension", arr2.ndim) # dimension of array
```

3. Array initialization methods

```python
# zeros array : all elements are 0
arr_zeros = np.zeros((2,3))
print(arr_zeros)
```

```python
# ones array : all elements are 1
arr_ones = np.ones((2,3))
print(arr_ones)
```

```python
# full array : all elements are same based on given value
arr_full = np.full((2,3),5)
print(arr_full)
```

```python
# Identity matrix : diagonal elements are 1
arr_identity = np.eye(3)
print(arr_identity)
```

```python
# empty array
print(np.empty(2))
```

```python
# evenly spaced array
arr_evenly_spaced = np.arange(2,10,2)
print(arr_evenly_spaced)
```

```python
# specific number of equally spaced values between start and end
arr_linspace = np.linspace(2,100,5)
print(arr_linspace)
```

```python
# random values array - float
arr_random = np.random.random((2,3))
print(arr_random)
```

```python
# random values array - integer
arr_random = np.random.randint(2,10,(2,3))
print(arr_random)
```

4. Array Indexing and Slicing

```python
a = np.array([1,3,5,7,9])
print(a)
```

```python
print(a[0]) # first element
print(a[4]) # last element
```

```python
print(a[-1]) # last element
```

```python
# slicing 
# array[start:end:step]
print(a[0:4:2]) # [start:end:step]
print(a[::2]) # [start:end:step]
print(a[::-1]) # reverse array
```

```python
# Indexing on 2D array
a = np.array([[1,3,5],[7,9,11]])
print(a)
```

```python
print(a[0]) # first row
print(a[1]) # second row
print(a[0][0]) # first element of first row
print(a[1][2]) # third element of second row
```

```python
print(a[:,0]) # first column
print(a[:,1]) # second column
print(a[:,2]) # third column
```

```python
# 3D Array: [height, rows,column]
a = np.array([[1,3,5],[7,9,11],[13,15,17]])
print(a)
```

```python
print(a[0])
print(a[:,0]) # first column
print(a[:,1]) # second column
print(a[:,2]) # third column
```
